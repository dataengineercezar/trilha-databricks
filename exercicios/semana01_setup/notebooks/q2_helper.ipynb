# q2_helper — Notebook auxiliar para demonstração de `%run`

Este notebook é executado pelo notebook principal via `%run ./q2_helper`.

**Como `%run` funciona:**
1. O Databricks executa este notebook **no mesmo contexto Python** do notebook chamador
2. Todas as variáveis e funções definidas aqui ficam disponíveis na sessão chamadora
3. É diferente de `import` — não há módulo, é execução direta de células

> Após o `%run`, o notebook principal pode usar `mensagem_helper` e `saudacao_q2()` como se fossem definidos lá.

```python
# Variáveis e funções exportadas para o notebook que chamar %run ./q2_helper
# Tudo definido aqui fica disponível no escopo do notebook chamador

# --- Constante ---
VERSAO_HELPER = "1.0"

# --- Variável simples ---
mensagem_helper = f"[q2_helper v{VERSAO_HELPER}] Notebook auxiliar carregado com sucesso via %run!"

# --- Função 1: saudação personalizada ---
def saudacao_q2(nome: str) -> str:
    """Retorna uma saudação com contexto de Databricks Serverless."""
    return f"Olá, {nome}! Executando no Databricks Serverless — variáveis compartilhadas via %run."

# --- Função 2: formata lista de datasets ---
def formatar_datasets(lista) -> str:
    """Recebe lista de FileInfo (dbutils.fs.ls) e retorna string formatada."""
    linhas = [f"  [{i+1:02d}] {item.name}" for i, item in enumerate(lista)]
    return "\n".join(linhas)

# Confirmar que o helper foi carregado (visível no output do %run)
print(mensagem_helper)
print(f"Funções disponíveis: saudacao_q2(), formatar_datasets()")
```
